```python
# Patriots vs Seahawks Defensive Scouting Model

# Goal:
# Analyze Seattle's defense and identify offensive opportunities for New England.

# Questions:
# - How efficient is Seattle's defense?
# - Which downs are vulnerable?
# - Which areas of the field give up value?
# - Which concepts should the Patriots attack?
```

```python
import nfl_data_py as nfl
import pandas as pd
import plotly.express as px

years = [2025]

pbp = nfl.import_pbp_data(years)

seattle = pbp[
    pbp["defteam"] == "SEA"
]

seattle = seattle[
    seattle["play_type"].isin(["pass","run"])
]

seattle.head()
```

```python
explosive = seattle[
    seattle["yards_gained"] >= 15
]


explosive_rate = (
    len(explosive) /
    len(seattle)
)


print(
    f"Explosive play rate allowed: {explosive_rate:.2%}"
)
```

```python
negative = seattle[
    seattle["epa"] < 0
]


negative_rate = (
    len(negative) /
    len(seattle)
)


print(
    f"Negative play rate: {negative_rate:.2%}"
)
```

```python
down_epa = (

seattle
.groupby("down")
["epa"]
.mean()
.reset_index()

)

down_epa
```

```python
fig = px.bar(
    down_epa,
    x="down",
    y="epa",
    title="Seattle EPA Allowed by Down"
)

fig.show()
```

```python
coverage = seattle[
    seattle["play_type"]=="pass"
]


coverage_metrics = {

"Completion %":
coverage.complete_pass.mean(),

"Avg EPA":
coverage.epa.mean(),

"Avg Air Yards":
coverage.air_yards.mean()

}


pd.DataFrame(
coverage_metrics,
index=[0]
)
```

```python
pressure = seattle[
    seattle["qb_hit"] == 1
]


print(
"QB hits:",
len(pressure)
)
```

```python
sacks = seattle[
    seattle["sack"] == 1
]

print(
"Sacks:",
len(sacks)
)
```

```python
route_defense = (

seattle
.groupby("route")
.agg(

EPA=("epa","mean"),
Attempts=("route","count"),
Yards=("yards_gained","mean")

)
.reset_index()

)


route_defense.sort_values(
"EPA"
)
```

```python
defense_score = {

"EPA Allowed":
seattle.epa.mean(),

"Explosive Rate":
explosive_rate,

"Negative Play Rate":
negative_rate,

"Success Allowed":
seattle.success.mean()

}


pd.DataFrame(
defense_score,
index=["Seattle"]
)
```

```python
patriots = pbp[
    (pbp["posteam"] == "NE") &
    (pbp["play_type"].isin(["pass","run"]))
]

patriots.head()
```

```python
seattle_offense = seattle.groupby(
    "play_type"
).agg(

    EPA=("epa","mean"),
    Yards=("yards_gained","mean"),
    Success=("success","mean")

).reset_index()


seattle_offense
```

```python
down_attack = (

seattle
.groupby("down")
.agg(

EPA=("epa","mean"),
Success=("success","mean")

)
.reset_index()

)

down_attack
```

```python
routes = (

patriots
.groupby("route")
.agg(

EPA=("epa","mean"),
Success=("success","mean"),
Targets=("route","count")

)
.reset_index()

)


routes.sort_values(
"EPA",
ascending=False
)
```

```python
attack_plan = routes.merge(
    route_defense,
    on="route",
    suffixes=("_NE","_SEA")
)


attack_plan["Matchup_Score"] = (

attack_plan["EPA_NE"]
*
attack_plan["EPA_SEA"]

)


attack_plan.sort_values(
"Matchup_Score",
ascending=False
)
```

```python
explosive_allowed = seattle[
    seattle["yards_gained"] >= 20
]


explosive_allowed.groupby(
"pass_location"
).size()
```

```python
attack_plan["Final_Attack_Score"] = (

attack_plan["Matchup_Score"]
*
attack_plan["Targets"]

)


attack_plan.sort_values(
"Final_Attack_Score",
ascending=False
).head(10)
```
